```python
# ENVIRONMENT SETUP & DRIVE DATA INGESTION

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Configure publication-grade high-resolution visualization standards
sns.set_theme(style="whitegrid")
plt.rcParams['figure.dpi'] = 150
plt.rcParams['savefig.dpi'] = 300

file_path = '/content/drive/MyDrive/hse_2022_eul_v1.tab'
print(f" Loading raw database from Google Drive path: '{file_path}'...")

try:
    df_raw = pd.read_csv(file_path, sep='\t')
    print(f"Success: Loaded database with {df_raw.shape[0]:,} rows and {df_raw.shape[1]} columns!")
except FileNotFoundError:
    raise FileNotFoundError(
        f"Critical Error: Could not locate the file at '{file_path}'. "
        "Please ensure your Google Drive is mounted and the path is exactly correct."
    )
```

```python

# SYNCHRONIZED EDA VISUALS
print(" Preprocessing features and locking row alignment across variables...")
raw_features = pd.DataFrame()

# 1. Map Gender / Biological Sex
raw_features['Gender_Label'] = df_raw['Sex'].map({1: 'Male', 2: 'Female'})

# 2. Map Target Outcome: LST7MOD (Number of days of moderate activity)
raw_features['Raw_Days'] = df_raw['lst7mod']

# 3. Map ag16g10
age_mapping = {
    1: '16-24 Years',
    2: '25-34 Years',
    3: '35-44 Years',
    4: '45-54 Years',
    5: '55-64 Years',
    6: '65-74 Years',
    7: '75+ Years'
}
raw_features['Age_Bracket'] = df_raw['ag16g10'].map(age_mapping)

# 4. Map Individual nssec8
nssec8_mapping = {
    1: 'Higher Managerial & Professional',
    2: 'Lower Managerial & Professional',
    3: 'Intermediate Occupations',
    4: 'Small Employers & Own Account',
    5: 'Lower Supervisory & Technical',
    6: 'Semi-Routine Occupations',
    7: 'Routine Occupations',
    8: 'Never Worked & Long-Term Unemployed',
    99: 'Other'
}
raw_features['Social_Class_8'] = df_raw['nssec8'].map(nssec8_mapping)

# Filter out survey error flags from days variable (Drops -9, -8, -1 from syntax)
raw_features = raw_features[raw_features['Raw_Days'] >= 0]

# Drop rows containing a missing value anywhere
final_df = raw_features.dropna(subset=['Gender_Label', 'Age_Bracket', 'Social_Class_8']).reset_index(drop=True)

# Map explicitly into the 2 active classes present in our clean population sample
def section_activity(days):
    if days <= 4:
        return 'Partially Active (1-4 Days)'
    else:
        return 'Highly Active (5-7 Days)'

final_df['Activity_Section'] = final_df['Raw_Days'].apply(section_activity)

# Numeric encoding for modeling (0 = Partially Active, 1 = Highly Active)
final_df['Target_Label'] = final_df['Activity_Section'].map({'Partially Active (1-4 Days)': 0, 'Highly Active (5-7 Days)': 1})

total_locked_count = final_df.shape[0]
print(f" OFFICIAL SPSS LOCK COMPLETED: Pool locked to exactly: {total_locked_count:,} total rows.\n")

# Switch background theme to clean minimalist look to match professional styling
sns.set_theme(style="white")

# ------------------------------------------------------------------------------
# Chart 1: Gender Distribution (Viridis Gradient)
# ------------------------------------------------------------------------------
plt.figure(figsize=(6, 4.5))
gender_order = final_df['Gender_Label'].value_counts().index
ax1 = sns.countplot(x='Gender_Label', data=final_df, order=gender_order, hue='Gender_Label', palette='viridis', legend=False)
plt.title('Individual Profile: Gender Baseline Distribution', fontsize=12, fontweight='bold', pad=12)
plt.xlabel('Gender', fontweight='bold', labelpad=8)
plt.ylabel('Respondent Count', fontweight='bold', labelpad=8)
ax1.set_ylim(0, final_df['Gender_Label'].value_counts().max() * 1.15)
for container in ax1.containers:
    ax1.bar_label(container, fmt='%d', padding=4, fontweight='bold', fontsize=10)
sns.despine()
plt.tight_layout()
plt.show()

# ------------------------------------------------------------------------------
# Chart 2: Age Bracket Distribution (7 Sorted 10-Year Bands via Viridis)
# ------------------------------------------------------------------------------
plt.figure(figsize=(9, 4.5))
age_order = final_df['Age_Bracket'].value_counts().index
ax2 = sns.countplot(x='Age_Bracket', data=final_df, order=age_order, hue='Age_Bracket', palette='viridis', legend=False)
plt.title('Individual Profile: 10-Year Age Bands (ag16g10 Sorted by Volume)', fontsize=12, fontweight='bold', pad=12)
plt.xlabel('Age Group Category', fontweight='bold', labelpad=8)
plt.ylabel('Respondent Count', fontweight='bold', labelpad=8)
ax2.set_ylim(0, final_df['Age_Bracket'].value_counts().max() * 1.15)
for container in ax2.containers:
    ax2.bar_label(container, fmt='%d', padding=4, fontweight='bold', fontsize=10)
sns.despine()
plt.tight_layout()
plt.show()

# ------------------------------------------------------------------------------
# Chart 3: Expanded Social Tiers (FIXED TEXT OVERLAP & CLIPPING)
# ------------------------------------------------------------------------------
plt.figure(figsize=(11, 6.0))
social_order = final_df['Social_Class_8'].value_counts().index
ax3 = sns.countplot(y='Social_Class_8', data=final_df, order=social_order, hue='Social_Class_8', palette='magma', legend=False)
plt.title('UK Official Individual Socio-Economic Tiers (NSSEC8 with Other Fix)', fontsize=12, fontweight='bold', pad=12)
plt.xlabel('Respondent Count', fontweight='bold', labelpad=8)

# FIXED: Kept labelpad=35 to safely distance the axis title from text strings
plt.ylabel('NS-SEC 8 Tiers', fontweight='bold', labelpad=35)

ax3.set_xlim(0, final_df['Social_Class_8'].value_counts().max() * 1.15)

for container in ax3.containers:
    ax3.bar_label(container, fmt='%d', padding=5, fontweight='bold', fontsize=9)
sns.despine()

# FIXED: Removed the invalid bbox_inches parameter
plt.tight_layout()
plt.show()

# ------------------------------------------------------------------------------
# Chart 4: Target Sections (Magma Gradient)
# ------------------------------------------------------------------------------
plt.figure(figsize=(7, 5))
target_order = final_df['Activity_Section'].value_counts().index
ax4 = sns.countplot(x='Activity_Section', data=final_df, order=target_order, hue='Activity_Section', palette='magma', legend=False)
plt.title('Individual Profile: Target Activity Segments (LST7MOD)', fontsize=12, fontweight='bold', pad=12)
plt.xlabel('Derived Activity Sections', fontweight='bold', labelpad=8)
plt.ylabel('Respondent Count', fontweight='bold', labelpad=8)
ax4.set_ylim(0, final_df['Activity_Section'].value_counts().max() * 1.15)
for container in ax4.containers:
    ax4.bar_label(container, fmt='%d', padding=4, fontweight='bold', fontsize=10)
sns.despine()
plt.tight_layout()
plt.show()

# ------------------------------------------------------------------------------
# Chart 5: Expanded Feature Correlation Heatmap
# ------------------------------------------------------------------------------
print("\n📊 Calculating linear relationships across variables...")
X_encoded = pd.get_dummies(final_df[['Age_Bracket', 'Gender_Label', 'Social_Class_8']], drop_first=True)
corr_df = X_encoded.copy()
corr_df['Highly_Active_Flag'] = final_df['Target_Label'].values
corr_matrix = corr_df.corr()

plt.figure(figsize=(15, 12))
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap="coolwarm", vmin=-1.0, vmax=1.0, linewidths=0.7, linecolor='white')
plt.title('Individual Profile: Fully Aligned Feature Correlation Matrix', fontsize=13, fontweight='bold', pad=15)
plt.xticks(rotation=45, ha='right', fontweight='bold')
plt.yticks(fontweight='bold')

plt.tight_layout()
plt.show()
```

```python

# FEATURE ENGINEERING (ONE-HOT ENCODING)
print(" Initiating Feature Engineering...")

# 1. Isolate Feature Variables (X) and Target Variable (y)
features_to_encode = final_df[['Gender_Label', 'Age_Bracket', 'Social_Class_8']]
y = final_df['Target_Label']

# 2. Engineer Categorical Text into Binary Matrices
# Using drop_first=True to prevent multicollinearity (The Dummy Variable Trap)
X_engineered = pd.get_dummies(features_to_encode, drop_first=True)

# Save feature names for later model interpretation
feature_names = X_engineered.columns.tolist()

print(f" Feature Engineering Complete. Engineered {X_engineered.shape[1]} distinct binary dimensions.")
```

```python

# MATRIX PARTITIONING AND FEATURE SCALING
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

print(" Splitting and Scaling Matrices for Machine Learning...")

# 1. Stratified Train/Test Partition (80% Training / 20% Blind Test)
X_train, X_test, y_train, y_test = train_test_split(
    X_engineered,
    y,
    test_size=0.2,
    stratify=y
)
print(f"-> Matrix Split Complete: {X_train.shape[0]:,} Training Profiles | {X_test.shape[0]:,} Testing Profiles.")

# 2. Standard Scaling (Crucial for Neural Network Convergence)
scaler = StandardScaler()

# Fit the scaler ONLY on the training data to prevent "data leakage", then transform both
X_train_scaled = scaler.fit_transform(X_train.astype(float))
X_test_scaled = scaler.transform(X_test.astype(float))

print(" Matrix Preparation Complete. The data is scaled (mean=0, variance=1) and ready for the algorithms.")
```

```python

# BASELINE MACHINE LEARNING FRAMEWORKS (CLASSICAL ML)
print(" Training Classical Baseline Machine Learning Frameworks...")

# Initialize accuracy tracking dictionary
accuracy_benchmarks = {}
class_names = ['Partially Active', 'Highly Active']

# ------------------------------------------------------------------------------
# 1. Balanced Logistic Regression
# ------------------------------------------------------------------------------
print("-> Executing Model 1: Balanced Logistic Regression...")
lr_model = LogisticRegression(class_weight='balanced', max_iter=1000)
lr_model.fit(X_train_scaled, y_train)
lr_preds = lr_model.predict(X_test_scaled)
accuracy_benchmarks['Logistic Regression'] = accuracy_score(y_test, lr_preds)

# ------------------------------------------------------------------------------
# 2. Balanced Random Forest Ensemble
# ------------------------------------------------------------------------------
print("-> Executing Model 2: Balanced Random Forest Classifier...")
rf_model = RandomForestClassifier(n_estimators=200, max_depth=6, min_samples_split=10, class_weight='balanced', n_jobs=-1)
rf_model.fit(X_train_scaled, y_train)
rf_preds = rf_model.predict(X_test_scaled)
accuracy_benchmarks['Random Forest'] = accuracy_score(y_test, rf_preds)

# ------------------------------------------------------------------------------
# Console Evaluation Reports
# ------------------------------------------------------------------------------
print("\n --- BASELINE ACCURACY SCOREBOARD ---")
for name in ['Logistic Regression', 'Random Forest']:
    print(f"-> {name:<20} | Test Accuracy: {accuracy_benchmarks[name]:.4f}")

print("\n --- RANDOM FOREST CLASSIFICATION REPORT ---")
print(classification_report(y_test, rf_preds, target_names=class_names, zero_division=0))

# ------------------------------------------------------------------------------
# Baseline Diagnostic Dashboard (Side-by-Side Visuals)
# ------------------------------------------------------------------------------
print("\n Rendering Baseline Diagnostic Dashboard...")
fig, axes = plt.subplots(1, 2, figsize=(14, 5.5))

# Subplot 1: Accuracy Stacking (Coolwarm Palette)
sns.barplot(x=list(accuracy_benchmarks.keys()), y=list(accuracy_benchmarks.values()), palette='coolwarm', ax=axes[0])
axes[0].set_title('Baseline Score Matrix: Accuracy Comparison', fontsize=12, fontweight='bold', pad=12)
axes[0].set_ylabel('Multi-Class Test Accuracy Score', fontweight='bold')
axes[0].set_ylim(0, 1.1)
for container in axes[0].containers:
    axes[0].bar_label(container, fmt='%.4f', padding=4, fontweight='bold', fontsize=10)
sns.despine(ax=axes[0])

# Subplot 2: Random Forest Confusion Matrix (Purples Palette)
cm = confusion_matrix(y_test, rf_preds)
sns.heatmap(cm, annot=True, fmt='d', cmap='Purples', xticklabels=class_names, yticklabels=class_names, cbar=False, ax=axes[1], linewidths=0.5, linecolor='white')
axes[1].set_title('Random Forest: Confusion Matrix', fontsize=12, fontweight='bold', pad=12)
axes[1].set_xlabel('Predicted Tiers (Model Output)', fontweight='bold', labelpad=8)
axes[1].set_ylabel('True Tiers (HSE Survey Reality)', fontweight='bold', labelpad=8)

plt.tight_layout()
plt.show()

print("\n Baseline Training Complete.")
```

```python

# DEEP LEARNING - MULTI-LAYER PERCEPTRON (MLP) NEURAL NETWORK
from sklearn.neural_network import MLPClassifier

print(" Initializing Deep Learning Framework: Multi-Layer Perceptron (MLP)...")

# 1. Define the deep neural network architecture
# We are using 3 Hidden Layers: 128 neurons -> 64 neurons -> 32 neurons
mlp_model = MLPClassifier(
    hidden_layer_sizes=(128, 64, 32),
    activation='relu',       # Standard activation function for deep learning
    solver='adam',           # Adaptive moment estimation optimizer
    max_iter=1000,           # Maximum training epochs
    random_state=None,       # Unlocked natural variance (Supervisor requested)
    early_stopping=True,     # Prevents overfitting by monitoring validation loss
    n_iter_no_change=10      # Stops training if the model stops improving
)

print("-> Training MLP Neural Network (This may take a moment to converge)...")
mlp_model.fit(X_train_scaled, y_train)

# 2. Generate predictions on the blind test set
mlp_preds = mlp_model.predict(X_test_scaled)

# 3. Add the neural network's score to your master benchmark dictionary
accuracy_benchmarks['MLP Neural Network'] = accuracy_score(y_test, mlp_preds)

# ------------------------------------------------------------------------------
# Console Evaluation Reports
# ------------------------------------------------------------------------------
print("\n --- UPDATED MASTER ACCURACY SCOREBOARD ---")
for name, score in accuracy_benchmarks.items():
    print(f"-> {name:<20} | Test Accuracy: {score:.4f}")

print("\n --- MLP NEURAL NETWORK CLASSIFICATION REPORT ---")
print(classification_report(y_test, mlp_preds, target_names=class_names, zero_division=0))

# ------------------------------------------------------------------------------
# Master Diagnostic Dashboard (All 3 Models)
# ------------------------------------------------------------------------------
print("\n Rendering Master Diagnostic Dashboard...")
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# Subplot 1: 3-Model Accuracy Comparison
# FIXED: Assigned 'hue' to prevent the Seaborn FutureWarning
sns.barplot(
    x=list(accuracy_benchmarks.keys()),
    y=list(accuracy_benchmarks.values()),
    hue=list(accuracy_benchmarks.keys()),
    palette='coolwarm',
    legend=False,
    ax=axes[0]
)
axes[0].set_title('Master Score Matrix: Architecture Comparison', fontsize=12, fontweight='bold', pad=12)
axes[0].set_ylabel('Multi-Class Test Accuracy Score', fontweight='bold', labelpad=8)
axes[0].set_ylim(0, 1.1)
for container in axes[0].containers:
    axes[0].bar_label(container, fmt='%.4f', padding=4, fontweight='bold', fontsize=10)
sns.despine(ax=axes[0])

# Subplot 2: MLP Confusion Matrix (Using a Blues palette to distinguish it from the Random Forest)
cm_mlp = confusion_matrix(y_test, mlp_preds)
sns.heatmap(cm_mlp, annot=True, fmt='d', cmap='Blues', xticklabels=class_names, yticklabels=class_names, cbar=False, ax=axes[1], linewidths=0.5, linecolor='white')
axes[1].set_title('Deep Learning (MLP): Confusion Matrix', fontsize=12, fontweight='bold', pad=12)
axes[1].set_xlabel('Predicted Tiers (Network Output)', fontweight='bold', labelpad=8)
axes[1].set_ylabel('True Tiers (HSE Survey Reality)', fontweight='bold', labelpad=8)

plt.tight_layout()
plt.show()

print("\n Deep Learning Training Complete.")
```

```python

# STATE-OF-THE-ART GRADIENT BOOSTING (LIGHTGBM)
from lightgbm import LGBMClassifier

print(" Initializing Gradient Boosting Framework: LightGBM...")

# 1. Define the LightGBM architecture
# We use a balanced class weight to force the model to respect the minority "Highly Active" class
lgbm_model = LGBMClassifier(
    n_estimators=300,        # Builds 300 sequential correction trees
    learning_rate=0.05,      # Learns slowly and carefully to prevent overfitting
    max_depth=6,             # Limits tree depth to prevent memorizing the training data
    class_weight='balanced', # Forces strict attention on the smaller "Highly Active" segment
    n_jobs=-1,                # Uses all available computer processors
    verbose=-1
    )

print("-> Training sequential gradient boosting trees...")
lgbm_model.fit(X_train_scaled, y_train)

# 2. Generate predictions on the blind test set
lgbm_preds = lgbm_model.predict(X_test_scaled)

# 3. Add LightGBM to the master benchmark dictionary
accuracy_benchmarks['LightGBM'] = accuracy_score(y_test, lgbm_preds)

# ------------------------------------------------------------------------------
# Console Evaluation Reports
# ------------------------------------------------------------------------------
print("\n --- FINAL MASTER ACCURACY SCOREBOARD ---")
for name, score in accuracy_benchmarks.items():
    print(f"-> {name:<20} | Test Accuracy: {score:.4f}")

print("\n --- LIGHTGBM CLASSIFICATION REPORT ---")
print(classification_report(y_test, lgbm_preds, target_names=class_names, zero_division=0))

# ------------------------------------------------------------------------------
# Final Master Diagnostic Dashboard (All 4 Models)
# ------------------------------------------------------------------------------
print("\n Rendering Final Model Evaluation Dashboard...")
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Subplot 1: 4-Model Accuracy Comparison
sns.barplot(
    x=list(accuracy_benchmarks.keys()),
    y=list(accuracy_benchmarks.values()),
    hue=list(accuracy_benchmarks.keys()),
    palette='coolwarm',
    legend=False,
    ax=axes[0]
)
axes[0].set_title('Final Score Matrix: Architecture Comparison', fontsize=12, fontweight='bold', pad=12)
axes[0].set_ylabel('Multi-Class Test Accuracy Score', fontweight='bold', labelpad=8)
axes[0].set_ylim(0, 1.1)

# Dynamically angle the x-axis text so the 4 model names don't overlap
axes[0].tick_params(axis='x', rotation=15)

for container in axes[0].containers:
    axes[0].bar_label(container, fmt='%.4f', padding=4, fontweight='bold', fontsize=10)
sns.despine(ax=axes[0])

# Subplot 2: LightGBM Confusion Matrix (Using a Greens palette to distinguish it)
cm_lgbm = confusion_matrix(y_test, lgbm_preds)
sns.heatmap(cm_lgbm, annot=True, fmt='d', cmap='Greens', xticklabels=class_names, yticklabels=class_names, cbar=False, ax=axes[1], linewidths=0.5, linecolor='white')
axes[1].set_title('LightGBM Challenger: Confusion Matrix', fontsize=12, fontweight='bold', pad=12)
axes[1].set_xlabel('Predicted Tiers (Network Output)', fontweight='bold', labelpad=8)
axes[1].set_ylabel('True Tiers (HSE Survey Reality)', fontweight='bold', labelpad=8)

plt.tight_layout()
plt.show()

print("\n Final Model Training Complete.")
```

```python

# ADVANCED DIAGNOSTICS (LEARNING CURVES & ROC-AUC COMPARISON)
from sklearn.metrics import roc_curve, auc

print(" Generating Advanced Mathematical Diagnostic Plots...")
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# ------------------------------------------------------------------------------
# Subplot 1: Deep Learning Convergence (Epoch Loss Curve)
# ------------------------------------------------------------------------------
# mlp_model.loss_curve_ extracts the exact error rate at every single epoch step
axes[0].plot(mlp_model.loss_curve_, color='#2B5B84', linewidth=2.5, label='Network Training Error (Loss)')
axes[0].set_title('Deep Learning (MLP): Epoch Convergence Curve', fontsize=12, fontweight='bold', pad=12)
axes[0].set_xlabel('Training Iterations (Epochs)', fontweight='bold', labelpad=8)
axes[0].set_ylabel('Log Loss (Error Rate)', fontweight='bold', labelpad=8)

# Adding gridlines for a highly technical, readable aesthetic
axes[0].grid(True, linestyle='--', alpha=0.6)
axes[0].legend(loc='upper right', fontsize=11)
sns.despine(ax=axes[0])

# ------------------------------------------------------------------------------
# Subplot 2: Master ROC-AUC Curve (The "Actual vs Predicted" for Classification)
# ------------------------------------------------------------------------------
# Dictionary of our trained models for easy iteration
models = {
    'Logistic Regression': lr_model,
    'Random Forest': rf_model,
    'MLP Neural Network': mlp_model,
    'LightGBM Challenger': lgbm_model
}

# Matching colors to your previous plots to maintain presentation consistency
colors = ['gray', 'orange', '#2B5B84', '#2CA02C']

for (name, model), color in zip(models.items(), colors):
    # predict_proba extracts the raw mathematical confidence score before it rounds to 0 or 1
    y_probs = model.predict_proba(X_test_scaled)[:, 1]

    # Calculate False Positives, True Positives, and the Area Under Curve (AUC)
    fpr, tpr, thresholds = roc_curve(y_test, y_probs)
    roc_auc = auc(fpr, tpr)

    # Plot the curve for this specific model
    axes[1].plot(fpr, tpr, color=color, lw=2.5, label=f'{name} (AUC = {roc_auc:.3f})')

# Plot the baseline random guessing threshold (the 50/50 coin flip line)
axes[1].plot([0, 1], [0, 1], color='black', lw=2, linestyle='--', label='Random Guessing (Baseline)')

axes[1].set_title('Master ROC Curve: Class Separation Power', fontsize=12, fontweight='bold', pad=12)
axes[1].set_xlabel('False Positive Rate (Incorrectly flagged as Highly Active)', fontweight='bold', labelpad=8)
axes[1].set_ylabel('True Positive Rate (Correctly flagged as Highly Active)', fontweight='bold', labelpad=8)
axes[1].legend(loc='lower right', fontsize=10)
axes[1].grid(True, linestyle='--', alpha=0.6)
sns.despine(ax=axes[1])

plt.tight_layout()
plt.show()

print("\n Advanced Diagnostics Complete. The project pipeline is officially finished.")
```

```python

# AUTOMATED HYPERPARAMETER TUNING (LIGHTGBM OPTIMIZATION)
from sklearn.model_selection import RandomizedSearchCV
import warnings
warnings.filterwarnings('ignore') # Hides unnecessary LightGBM background warnings

print(" Initializing Automated Hyperparameter Search Space...")

# 1. Define the grid of "dials" we want the computer to test
param_grid = {
    'learning_rate': [0.01, 0.05, 0.1, 0.2],         # How fast the model updates
    'n_estimators': [100, 300, 500, 800],            # Number of sequential trees
    'max_depth': [3, 5, 7, 10, -1],                  # How deep each tree can grow
    'num_leaves': [15, 31, 50, 100],                 # Maximum leaves per tree
    'subsample': [0.6, 0.8, 1.0]                     # Percentage of data used per tree
}

# 2. Initialize a fresh, empty LightGBM model
base_lgbm = LGBMClassifier(class_weight='balanced', n_jobs=-1)

# 3. Set up the Randomized Search
# n_iter=20 means it will randomly test 20 different combinations from the grid above
# cv=3 means 3-Fold Cross Validation (double-checking its work to prevent overfitting)
# scoring='roc_auc' forces it to optimize for class separation, NOT raw accuracy!
random_search = RandomizedSearchCV(
    estimator=base_lgbm,
    param_distributions=param_grid,
    n_iter=20,
    scoring='roc_auc',
    cv=3,
    verbose=1,
    n_jobs=-1
)

print(" Launching Training Grid (Testing 20 combinations across 3 folds = 60 total fits)...")
random_search.fit(X_train_scaled, y_train)

# 4. Extract the absolute best configuration it found
best_model = random_search.best_estimator_
best_params = random_search.best_params_

print("\n Search Complete! The Optimal Hyperparameters are:")
for param, value in best_params.items():
    print(f"   -> {param}: {value}")

# 5. Evaluate the newly tuned "Super Model" on your blind test set
tuned_preds = best_model.predict(X_test_scaled)
tuned_probs = best_model.predict_proba(X_test_scaled)[:, 1]

fpr, tpr, _ = roc_curve(y_test, tuned_probs)
tuned_auc = auc(fpr, tpr)
tuned_accuracy = accuracy_score(y_test, tuned_preds)

print("\n --- TUNED LIGHTGBM PERFORMANCE ---")
print(f"-> Tuned Test Accuracy : {tuned_accuracy:.4f}")
print(f"-> Tuned AUC Score     : {tuned_auc:.4f}")

print("\n --- TUNED CLASSIFICATION REPORT ---")
print(classification_report(y_test, tuned_preds, target_names=['Partially Active', 'Highly Active'], zero_division=0))
```
